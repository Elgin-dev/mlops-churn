```python
import pandas as pd

raw_path = "../data/raw/churn.csv"

df = pd.read_csv(raw_path)

print("Shape:", df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
print("Shape after removing duplicates:", df.shape)
print("Duplicates remaining:", df.duplicated().sum())
```

```python
numeric_cols = [
    "age",
    "tenure_months",
    "monthly_charges",
    "support_calls",
    "usage_hours"
]
```

```python
df["age"] = df["age"].fillna(df["age"].median())

df["monthly_charges"] = df["monthly_charges"].fillna(
    df["monthly_charges"].median()
)
```

```python
df[numeric_cols].isnull().sum()
```

```python
df["contract_type"].value_counts()
```

```python
df["contract_type"] = df["contract_type"].fillna(
    df["contract_type"].mode()[0]
)
```

```python
df["contract_type"].isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.iloc[df["monthly_charges"] > 200]
```

```python
import numpy as np
df.loc[
    (df["monthly_charges"] < 0) |
    (df["monthly_charges"] > 200),
    "monthly_charges"
] = np.nan
```

```python
df["monthly_charges"].isnull().sum()
```

```python
df["monthly_charges"] = df["monthly_charges"].fillna(
    df["monthly_charges"].median()
)
```

```python
print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicates:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
from pathlib import Path

processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)

df.to_csv(processed_path / "cleaned_data.csv", index=False)

print("Cleaned data saved successfully!")
print("Path:", processed_path / "cleaned_data.csv")
print("Shape:", df.shape)
```
